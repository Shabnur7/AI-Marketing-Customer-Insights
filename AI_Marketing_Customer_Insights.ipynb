```python
import pandas as pd

# Sample product review data
data = {
    "product": [
        "Wireless Earbuds", "Wireless Earbuds", "Wireless Earbuds",
        "Wireless Earbuds", "Wireless Earbuds",
        "Wireless Earbuds", "Wireless Earbuds", "Wireless Earbuds"
    ],
    "rating": [5, 4, 5, 3, 2, 5, 4, 3],
    "review": [
        "The sound quality is amazing and the battery lasts all day.",
        "Very comfortable and easy to connect to my phone.",
        "Great sound for the price. I use them every day.",
        "Sound is good but the microphone could be better.",
        "Battery life is disappointing. It dies too quickly.",
        "I love how lightweight these earbuds are.",
        "Good value but the charging case feels cheap.",
        "The sound is clear, but they sometimes disconnect."
    ]
}

df = pd.DataFrame(data)

print("Number of reviews:", len(df))
display(df)
```

```python
# Basic review analysis

print("Average rating:", round(df["rating"].mean(), 2))
print("\nRating distribution:")
print(df["rating"].value_counts().sort_index())

print("\nLowest-rated reviews:")
display(df[df["rating"] <= 3][["rating", "review"]])
```

```python
# Identify common themes from customer reviews

keywords = [
    "sound",
    "battery",
    "comfort",
    "microphone",
    "connection",
    "price",
    "quality",
    "lightweight"
]

theme_counts = {}

for keyword in keywords:
    count = df["review"].str.lower().str.contains(keyword).sum()
    theme_counts[keyword] = count

theme_df = pd.DataFrame(
    list(theme_counts.items()),
    columns=["Theme", "Mentions"]
).sort_values("Mentions", ascending=False)

display(theme_df)
```

```python
# Convert review themes into simple marketing insights

top_themes = theme_df[theme_df["Mentions"] > 0].head(5)

print("TOP CUSTOMER THEMES")
print("===================")

for _, row in top_themes.iterrows():
    theme = row["Theme"]
    mentions = row["Mentions"]

    if theme == "sound":
        insight = "Sound quality is a major customer interest."
    elif theme == "battery":
        insight = "Battery life is an important purchase consideration."
    elif theme == "comfort":
        insight = "Comfort and wearability matter to customers."
    elif theme == "microphone":
        insight = j"Call and microphone performance may influence satisfaction."
    elif theme == "connection":
        insight = "Connection stability can affect customer experience."
    else:
        insight = f"{theme.capitalize()} appears in customer feedback."

    print(f"\n{theme.upper()} ({mentions} mentions)")
    print(insight)
```

```python
# Generate marketing strategy from customer insights

marketing_strategy = {
    "Primary Target": "Customers who prioritize high-quality sound in their everyday listening experience.",
    "Secondary Target": "Customers looking for convenient earbuds with good battery life.",
    "Main Marketing Angle": "High-quality sound designed for everyday use.",
    "Supporting Angle": "Long-lasting battery for uninterrupted listening.",
    "Customer Need": "Customers want good sound quality without sacrificing convenience and comfort.",
    "Content Idea": "Create short-form content showing everyday situations where clear sound and reliable battery life matter."
}

print("AI MARKETING STRATEGY")
print("=====================")

for key, value in marketing_strategy.items():
    print(f"\n{key}:")
    print(value)
```

```python
# Prepare the customer feedback for AI analysis

reviews_text = "\n".join(
    [f"Rating: {row.rating}/5 - {row.review}" for _, row in df.iterrows()]
)

print(reviews_text)
```

```python
prompt = f"""
You are an AI marketing analyst.

Analyze the following customer reviews for a wireless earbuds product.

{reviews_text}

Identify:

1. Top 3 product strengths
2. Top 3 customer complaints
3. Customer needs
4. Marketing opportunities
5. Product improvement opportunities

Give a concise, business-focused analysis.
"""

print(prompt)
```

```python
# Step 8: Identify important customer themes

themes = {
    "Sound Quality": ["sound", "audio", "clear"],
    "Battery Life": ["battery", "dies"],
    "Comfort & Weight": ["comfortable", "lightweight"],
    "Connectivity": ["connect", "disconnect"],
    "Microphone": ["microphone"],
    "Value for Money": ["price", "value"],
    "Charging Case": ["charging case", "case"]
}

theme_results = []

for theme, keywords in themes.items():
    count = 0
    matching_reviews = []

    for review in df["review"]:
        review_lower = review.lower()

        if any(keyword in review_lower for keyword in keywords):
            count += 1
            matching_reviews.append(review)

    theme_results.append({
        "Theme": theme,
        "Mentions": count,
        "Example Reviews": " | ".join(matching_reviews)
    })

theme_df = pd.DataFrame(theme_results)

theme_df = theme_df.sort_values(
    by="Mentions",
    ascending=False
).reset_index(drop=True)

theme_df
```

```python
# Step 9: Calculate the importance of each customer theme

total_reviews = len(df)

theme_df["Percentage"] = (
    theme_df["Mentions"] / total_reviews * 100
).round(1)

theme_df
```

```python
# Step 10: Improve customer theme detection

themes = {
    "Sound Quality": [
        "sound", "audio", "clear"
    ],

    "Battery Life": [
        "battery", "dies", "lasts"
    ],

    "Comfort & Weight": [
        "comfortable", "lightweight"
    ],

    "Connectivity": [
        "connect", "disconnect"
    ],

    "Microphone": [
        "microphone"
    ],

    "Value for Money": [
        "price", "value", "cheap"
    ],

    "Charging Case": [
        "charging case", "case"
    ]
}

theme_results = []

for theme, keywords in themes.items():
    count = 0
    matching_reviews = []

    for review in df["review"]:
        review_lower = review.lower()

        if any(keyword in review_lower for keyword in keywords):
            count += 1
            matching_reviews.append(review)

    theme_results.append({
        "Theme": theme,
        "Mentions": count,
        "Percentage": round((count / len(df)) * 100, 1),
        "Example Reviews": " | ".join(matching_reviews)
    })

theme_df = pd.DataFrame(theme_results)

theme_df = theme_df.sort_values(
    by="Mentions",
    ascending=False
).reset_index(drop=True)

theme_df
```

```python
# Step 11: Identify customer strengths and pain points

strength_keywords = [
    "sound",
    "comfortable",
    "lightweight",
    "value",
    "price"
]

pain_point_keywords = [
    "battery",
    "microphone",
    "disconnect",
    "cheap"
]

strengths = []
pain_points = []

for _, row in df.iterrows():
    review = row["review"].lower()

    if any(keyword in review for keyword in strength_keywords):
        strengths.append(row["review"])

    if any(keyword in review for keyword in pain_point_keywords):
        pain_points.append(row["review"])

print("CUSTOMER STRENGTHS")
print("=" * 40)

for review in strengths:
    print("-", review)

print("\nCUSTOMER PAIN POINTS")
print("=" * 40)

for review in pain_points:
    print("-", review)
```

```python
# Step 12: Create a clean customer insight table

def classify_customer_feedback(rating):
    if rating >= 4:
        return "Strength"
    elif rating == 3:
        return "Needs Improvement"
    else:
        return "Pain Point"

df["customer_insight"] = df["rating"].apply(classify_customer_feedback)

insight_table = df[
    ["review_id", "rating", "review", "sentiment", "customer_insight"]
].copy()

insight_table
```

```python
# Step 12: Create a clean customer insight table Corrected

def classify_customer_feedback(rating):
    if rating >= 4:
        return "Strength"
    elif rating == 3:
        return "Needs Improvement"
    else:
        return "Pain Point"

# Create customer insight classification
df["customer_insight"] = df["rating"].apply(classify_customer_feedback)

# Create a simple review ID if it doesn't already exist
df["review_id"] = range(1, len(df) + 1)

# Create sentiment from the rating
def classify_sentiment(rating):
    if rating >= 4:
        return "Positive"
    elif rating == 3:
        return "Neutral"
    else:
        return "Negative"

df["sentiment"] = df["rating"].apply(classify_sentiment)

# Create the final insight table
insight_table = df[
    ["review_id", "rating", "review", "sentiment", "customer_insight"]
].copy()

insight_table
```

```python
# Step 13: Calculate customer insight percentages

total_reviews = len(df)

strength_count = (df["customer_insight"] == "Strength").sum()
improvement_count = (df["customer_insight"] == "Needs Improvement").sum()
pain_count = (df["customer_insight"] == "Pain Point").sum()

strength_percentage = round((strength_count / total_reviews) * 100, 1)
improvement_percentage = round((improvement_count / total_reviews) * 100, 1)
pain_percentage = round((pain_count / total_reviews) * 100, 1)

print("CUSTOMER INSIGHT SUMMARY")
print("=" * 40)

print(f"Total Reviews: {total_reviews}")
print(f"Average Rating: {df['rating'].mean():.2f}/5")

print(f"\nStrengths: {strength_count} ({strength_percentage}%)")
print(f"Needs Improvement: {improvement_count} ({improvement_percentage}%)")
print(f"Pain Points: {pain_count} ({pain_percentage}%)")
```

```python
# Step 14: Identify marketing opportunities

marketing_opportunities = []

for _, row in theme_df.iterrows():
    theme = row["Theme"]
    mentions = row["Mentions"]
    percentage = row["Percentage"]

    if theme in ["Sound Quality", "Comfort & Weight", "Value for Money"]:
        opportunity = "Use as a key marketing message"

    elif theme in ["Battery Life", "Connectivity", "Microphone", "Charging Case"]:
        opportunity = "Address as a product improvement or reassurance point"

    else:
        opportunity = "Monitor customer feedback"

    marketing_opportunities.append({
        "Theme": theme,
        "Customer Mentions": mentions,
        "Review Percentage": percentage,
        "Marketing Recommendation": opportunity
    })

marketing_df = pd.DataFrame(marketing_opportunities)

marketing_df
```

```python
# Step 15: Generate a data-driven marketing strategy

top_marketing_themes = marketing_df[
    marketing_df["Marketing Recommendation"] == "Use as a key marketing message"
].copy()

improvement_themes = marketing_df[
    marketing_df["Marketing Recommendation"] == "Address as a product improvement or reassurance point"
].copy()

print("DATA-DRIVEN MARKETING STRATEGY")
print("=" * 50)

print("\n1. CORE MARKETING MESSAGES")
print("-" * 30)

for _, row in top_marketing_themes.iterrows():
    print(
        f"• {row['Theme']} "
        f"({row['Review Percentage']}% of reviews)"
    )

print("\n2. CUSTOMER CONCERNS TO ADDRESS")
print("-" * 30)

for _, row in improvement_themes.iterrows():
    print(
        f"• {row['Theme']} "
        f"({row['Review Percentage']}% of reviews)"
    )

print("\n3. RECOMMENDED MARKETING APPROACH")
print("-" * 30)

print(
    "Emphasize sound quality, comfort, lightweight design, "
    "and value for money in marketing campaigns. "
    "At the same time, avoid making unsupported claims about "
    "battery life, connectivity, microphone performance, or "
    "charging-case quality."
)
```
